###Does delivery performance affect customer satisfaction/review scores?

```python
output_review_delivery_path = 's3://ims-ecom-stage/review_delivery/'
```

```python
review_delivery_df = spark.sql("""
SELECT CASE
            WHEN o.late_delivery_flag = true THEN 'late'
            WHEN o.late_delivery_flag = false THEN 'on time'
            WHEN o.late_delivery_flag IS NULL THEN 'unknown'
        END AS delivery_status,
        AVG(r.review_score) AS avg_review_score,
        COUNT(r.review_score) AS num_reviews
    FROM ecommerce.staging.orders o join ecommerce.staging.reviews r
    ON o.order_id = r.order_id
    GROUP BY late_delivery_flag
""")
```

```python
review_delivery_df.write \
    .format('delta')  \
    .mode("overwrite") \
    .save(output_review_delivery_path)
```

```python
  %sql
CREATE TABLE IF NOT EXISTS ecommerce.staging.review_delivery
USING delta
LOCATION 's3://ims-ecom-stage/review_delivery/';
```
