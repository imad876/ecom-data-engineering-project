```python
import pyspark.sql.functions as F


```

```python
input_sellers_path = 's3://ims-ecom-raw/sellers/'
output_sellers_path = 's3://ims-ecom-stage/sellers/'
```

```python
sellers_df = spark.read.format("delta").load(input_sellers_path)
```

```python
sellers_df.write \
  .format("delta") \
  .mode("overwrite") \
  .save(output_sellers_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecommerce.staging.sellers
USING DELTA
LOCATION 's3://ims-ecom-stage/sellers/';
```
