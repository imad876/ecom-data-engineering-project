```python
import pyspark.sql.functions as F


```

```python
input_reviews_path = 's3://ims-ecom-raw/reviews/'
output_reviews_path = 's3://ims-ecom-stage/reviews/'
```

```python
reviews_df = spark.read.format("delta").load(input_reviews_path)
```

```python
reviews_df = reviews_df \
    .withColumn("has_review_comment", F.when(F.col("review_comment_message").isNotNull(), 1).otherwise(0)
)
```

```python
reviews_df.write \
  .format("delta") \
  .mode("overwrite") \
  .save(output_reviews_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecommerce.staging.reviews
USING DELTA
LOCATION 's3://ims-ecom-stage/reviews/';
```
