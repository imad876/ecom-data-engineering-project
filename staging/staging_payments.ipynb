```python
import pyspark.sql.functions as F


```

```python
input_payments_path = 's3://ims-ecom-raw/payments/'
output_payments_path = 's3://ims-ecom-stage/payments/'
```

```python
payments_df = spark.read.format("delta").load(input_payments_path)
```

```python
payments_df.write \
  .format("delta") \
  .mode("overwrite") \
  .save(output_payments_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecommerce.staging.payments
USING DELTA
LOCATION 's3://ims-ecom-stage/payments/';
```
