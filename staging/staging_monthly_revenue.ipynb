```python
output_monthly_revenue_path = 's3://ims-ecom-stage/monthly_revenue/'
```

```python
monthly_revenue_df = spark.sql("""
select o.purchase_month, o.purchase_year, p.product_category_name, c.customer_state, sum(oi.price) as total_revenue
from ecommerce.staging.orders o 
join ecommerce.staging.order_items oi on o.order_id = oi.order_id 
join ecommerce.staging.products p on oi.product_id = p.product_id 
join ecommerce.staging.customers c on o.customer_id = c.customer_id

GROUP BY
    purchase_month,
    purchase_year,
    product_category_name,
    customer_state
ORDER BY
    purchase_year,
    purchase_month,
    customer_state,
    product_category_name
""")
```

```python
monthly_revenue_df.write \
    .format('delta')  \
    .mode("overwrite") \
    .save(output_monthly_revenue_path)
```

```python
  %sql
CREATE TABLE IF NOT EXISTS ecommerce.staging.monthly_revenue
USING delta
LOCATION 's3://ims-ecom-stage/monthly_revenue/';
```
