```python
import pyspark.sql.functions as F
```

```python
source_path = 's3://ims-ecom-landing/olist/reviews/'
output_path = 's3://ims-ecom-raw/reviews/'
  
```

```python
if check_file_exists(source_path):
    df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("quote", '"') \
    .option("escape", '"') \
    .option("multiLine", True) \
    .csv(source_path)
    
    display(df)
    df.printSchema()

else:
    dbutils.notebook.exit("File not found, stopping execution.")
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(output_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecommerce.raw.reviews
USING DELTA
LOCATION 's3://ims-ecom-raw/reviews/';
```
