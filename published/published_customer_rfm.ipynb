```python
output_customer_rfm_path = 's3://ims-ecom-published/customer_rfm/'

customer_rfm_df = spark.read.table("ecommerce.staging.customer_rfm")

customer_rfm_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save(output_customer_rfm_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecommerce.published.customer_rfm
USING DELTA
LOCATION 's3://ims-ecom-published/customer_rfm/';
```

```python
%sql
SELECT * FROM ecommerce.published.customer_rfm LIMIT 5
```

```python
%sql
SELECT frequency, COUNT(*) AS customers
FROM ecommerce.published.customer_rfm
GROUP BY frequency
ORDER BY frequency
```

```python
%sql
SELECT customer_segment, COUNT(*) AS customers
FROM ecommerce.published.customer_rfm
GROUP BY customer_segment
ORDER BY customers DESC
```
